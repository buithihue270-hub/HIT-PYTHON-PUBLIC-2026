# Function

### Định nghĩa
- Hàm là một khối mã chỉ chạy khi nó được gọi
- Có thể truyền dữ liệu, được gọi là tham số vào một hàm
- Kết quả là hàm có thể trả về dữ liệu


## Khai báo

````python
# Sử dụng từ khóa def để định nghĩa hàm
def hello():
    print("Hello World")

# Gọi hàm
hello()
````

## Đối số

````python
## Đối số được đặt trong cặp ngoặc
def hello(name):
    print("Xin chào " + name)

# Gọi hàm truyền tham số
hello("Nguyễn Văn An")
hello("Trần Thị Bình")
````

````python
# Nhiều đối số, mỗi đối số cách nhau bởi dấu phẩy
def hello(last_name, middle_name, first_name):
    print("Xin chào " + last_name + " " + middle_name + " " + first_name)

hello("Nguyễn", "Văn", "An")
````

````python
# Khi không xác định được số đối số, chúng ta có thể dùng *
def timetable(*subject):
    print("Môn 1: " + subject[0])
    print("Môn 2: " + subject[1])

timetable("Toán", "Lý", "Hóa", "Sinh", "Văn", "Sử", "Địa")
````

````python
def calculate_sum(*value):
    sum = 0
    for item in value:
        sum += item
    print(sum)

calculate_sum(1, 2)
calculate_sum(1, 2, 3, 4, 5)
````

````python
# Truyền nhiều đối số, xác định thông qua tên, sử dụng **
def hello(**name):
    print("Xin chào " + name["last_name"])

hello(first_name="Bình", middle_name="Văn", last_name="Nguyễn")
hello(last_name="Tùng", middle_name="Văn", first_name="Lê")
````

## Trả về

````python
# Sử dụng từ khóa "return" để trả về gía trị
def calculate_product(*value):
    result = 1
    for item in value:
        result *= item
    return result

result1 = calculate_product(1, 4, 6)
result2 = calculate_product(7, 4, 6)
print(result1)
print(result2)
````

# Bài tập 2
Tìm ước chung lớn nhất của 2 số a, b

Xây dựng hàm: gcd(a, b) => Trả về ước chung lớn nhất

**Ví dụ:** 1, 13 => gcd(1, 13) => 1

**Ví dụ:** 35, 77 => gcd(35, 77) => 7

**Gợi ý:**

- Thuật toán đơn:
```
35, 42
35, 7
28, 7
21, 7
14, 7
7, 7
```
