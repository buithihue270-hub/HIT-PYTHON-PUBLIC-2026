# Lambda Function

Lambda là một hàm ẩn danh nhỏ

Một hàm Lambda có thể nhận bất kỳ đối số nào, nhưng chỉ có thể có 1 biểu thức

### 1) Cú pháp: `lambda arguments: expression`

```python
isEven = lambda x : x % 2 == 0
print(isEven(5))
print(isEven(8))
```

```python
calculate_sum = lambda a, b : a + b
print(calculate_sum(5, 8))
print(calculate_sum(5, 1))
```

```python
# Sử dụng lambda function bên trong function
def square(x):
    return x ** 2

def exp_3(x):
    return x ** 3

def exp(n):
    return lambda x: x ** n

hamMu2 = exp(2)
hamMu2b = lambda x : x ** 2
hamMu3 = exp(3)
hamMu4 = exp(4)

print(hamMu2(3))
print(hamMu3(3))
print(hamMu4(3))
```

### 2) `map`, `filter`, `reduce`

- `map(fn, iterable)` áp dụng fn lên từng phần tử. thực hiện thay đổi trên toàn bộ danh sách 
- `filter(fn, iterable)` lọc phần tử sao cho fn(item) là True.
- `reduce` nằm trong `functools` dùng để giảm một iterable về một giá trị duy nhất. Tính toán tổng,.. trên 1 danh sách

```python
from functools import reduce # thêm để sử dụng hàm reduce

nums = [1,2,3,4,5]
print(list(map(lambda x: x*2, nums)))
print(list(filter(lambda x: x%2==0, nums)))
print(reduce(lambda a,b: a+b, nums))
```

### 3) `sorted(..., key=...)` với `lambda`

Sử dụng lambda để chỉ định tiêu chí sắp xếp:

```python
pairs = [('a', 3), ('b', 1), ('c', 2)]# chỉ dùng được cho 2 phần tử
print(sorted(pairs, key=lambda x: x[1]))  # sắp theo phần tử thứ 2
```

### 4) Khi nào không nên dùng `lambda`

- Khi logic phức tạp: nên dùng `def` để dễ đọc và debug.
