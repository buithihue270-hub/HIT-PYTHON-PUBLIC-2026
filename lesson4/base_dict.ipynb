# Dictionary

Từ điển được sử dụng để lưu trữ các giá trị trong các cặp `key:value`

Từ điển là một tập hợp được sắp xếp có thứ tự, có thể thay đổi và không cho phép trùng lặp

Từ điển không cho phép có 2 mục cùng 1 khóa

````python
student = {
    "name": "Nguyen Van A",
    "age": 19,
    "gpa": 8.5
}

print(student)
````

## Truy cập các mục

````python
student = {
    "name": "Nguyen Van A",
    "age": 19,
    "gpa": 8.5
}

print(student["name"])
````

````python
# Sử dụng get() để lấy giá trị
student.get("name")
````

## Thay đổi giá trị

````python
student["name"] = "Trần Thị B"

print(student)
````

````python
# Sử dụng phương thức update()
student2 = {
    "name": "Lê Văn C",
    "age": 20,
    "gpa": 8.5
}

student.update({"name": "Nguyễn Hà My", "gpa": 9.0})
print(student)
````

## Thêm các mục

````python
student["yob"] = 2007 # Keyword mới
print(student)
````

````python
student.update({"hometown": "Hà Nội"})
print(student)
````

***Chú ý:*** Cú pháp giống thêm một giá trị mới, tuy nhiên khóa phải chưa tồn tại trong dictionary ban đầu

## Loại bỏ các mục

````python
# Phương thức pop()
student.pop("hometown")
print(student)
````

````python
# Phương thức popitem()
student.popitem()
print(student)
````

````python
# Từ khóa del xóa cặp key:value được chỉ định
del student["gpa"]
print(student)

# Từ khóa del dùng để xóa toàn bộ dict
del student
print(student)
````

````python
# Từ khóa clear để xóa toàn dict
student = {
    "name": "Nguyen Van A",
    "age": 19,
    "gpa": 8.5
}
print(student)
student.clear()
print(student)
````

## Duyệt dict bằng vòng lặp

````python
student = {
    "name": "Nguyen Van A",
    "age": 19,
    "gpa": 8.5
}

# In các khóa có trong từ điển
for x in student:
    print(x)

# Duyệt các giá trị
for x in student.values():
    print(x)

# Duyệt các khóa
for x in student.keys():
    print(x)

# Duyệt các cặp khóa - giá trị
for x, y in student.items():
    print(x, y)
````
