### Bài 1: Tính số ngày trong tháng

**Yêu cầu:** Cho biết tháng và năm, hãy in ra số ngày tương ứng có trong tháng đó.

**Lưu ý:**
- Tháng 1, 3, 5, 7, 8, 10, 12 có 31 ngày.
- Tháng 4, 6, 9, 11 có 30 ngày.
- Tháng 2 có 28 ngày, riêng năm nhuận có 29 ngày.
- Cách kiểm tra năm nhuận: Năm chia hết cho 400 hoặc (chia hết cho 4 và không chia hết cho 100).

**Ví dụ:**
1. **Input:** Tháng 10, Năm 2020 → **Output:** 31
2. **Input:** Tháng 2, Năm 2024 → **Output:** 29
3. **Input:** Tháng 2, Năm 2025 → **Output:** 28
4. **Input:** Tháng 11, Năm 1921 → **Output:** 30

```python
print("TÍNH SỐ NGÀY TRONG THÁNG")
thang = int(input("Nhập tháng:"))
nam = int(input("Nhập năm:"))
hop_le = True
if thang < 1 or thang > 12:
 hop_le = False
if hop_le == True:
  if thang == 1 or thang == 3 or thang == 5 or thang == 7 or thang == 8 or thang == 10 or thang == 12:
    print("Output: 31 ngày")
  elif thang == 4 or thang == 6 or thang == 9 or thang == 11:
    print("Output: 30 ngày")
  else:
    if nam % 400 == 0:
     print("Output: 29 ngày")
    elif nam % 4 == 0 and nam % 100 != 0:
     print("Output: 29 ngày")
    else:
      print("Output: 28 ngày")
else:
  print("Tháng nhập vào không hợp lệ.")
```

### Bài 3: Đếm chữ số, tính tổng chữ số và kiểm tra số nguyên tố

**Yêu cầu:** Nhập vào một số nguyên $n$, thực hiện các yêu cầu sau:
1.  Đếm số lượng chữ số của $n$.
2.  Tính tổng các chữ số của $n$.
3.  Kiểm tra $n$ có phải là số nguyên tố hay không.
4.  In các kết quả ra màn hình.

**Ví dụ:**
1. **Input:** 54 → **Output:** Số chữ số: 2, Tổng chữ số: 9, 54 không phải là số nguyên tố.
2. **Input:** 53 → **Output:** Số chữ số: 2, Tổng chữ số: 8, 53 là số nguyên tố.
3. **Input:** 123 → **Output:** Số chữ số: 3, Tổng chữ số: 6, 123 không phải là số nguyên tố.
4. **Input:** 2 → **Output:** Số chữ số: 1, Tổng chữ số: 2, 2 là số nguyên tố.

```python
print("ĐẾM CHỮ SỐ, TÍNH TỔNG CHỮ SỐ VÀ KIỂM TRA SỐ NGUYÊN TỐ")
n = int(input("Nhập số nguyên n:"))
so_tam = n
if so_tam < 0:
  so_tam = -so_tam
dem_chu_so = 0
tong_chu_so = 0
if so_tam == 0:
  dem_chu_so = 1
  tong_chu_so = 0
else:
  while so_tam > 0:
    chu_so = so_tam % 10
    tong_chu_so = tong_chu_so + chu_so
    dem_chu_so = dem_chu_so + 1
    so_tam = so_tam // 10
la_nguyen_to = True
if n < 2:
  la_nguyen_to = False
else:
  i = 2
  while i * i <= n:
    if n % i == 0:
      la_nguyen_to = False
    i = i +1
if la_nguyen_to == True:
  print("Số chữ số:", dem_chu_so, ", Tổng chữ số:", tong_chu_so, ",", n, "là số nguyên tố.")
else:
  print("Số chữ số:", dem_chu_so, ", Tổng chữ số:", tong_chu_so, ",", n, "không phải là số nguyên tố.")
```

### Bài 4: Bài toán đổi vỏ chai bia

**Yêu cầu:** Tại Hà Nội, các cửa hàng bia có chương trình khuyến mãi: Cứ **3 vỏ chai bia** sẽ đổi được **1 chai bia mới**. Biết rằng mỗi chai bia có giá **28 xu**. Với số tiền $N$ xu cho trước, hãy xác định tổng số chai bia tối đa bạn có thể uống được (bao gồm số chai mua bằng tiền và số chai đổi được từ vỏ).

**Ví dụ
:**
- **Input:** 28 xu
- **Output:** Số chai bia có thể mua được là: 1

- **Input:** 56 xu
- **Output:** Số chai bia có thể mua được là: 2

- **Input:** 84 xu
- **Output:** Số chai bia có thể mua được là: 4 (Mua 3 chai ban đầu, dư 3 vỏ đổi thêm được 1 chai nữa)

```python
print("ĐỔI VỎ CHIA BIA")
n = int(input(("Nhập số tiền: ")))
gia_chai = 28
so_chai_mua = n // gia_chai
tong_so_chai = so_chai_mua
so_vo = so_chai_mua
while so_vo >= 3:
  so_vo = so_vo - 3
  tong_so_chai = tong_so_chai + 1
  so_vo = so_vo + 1
print("Output: Số chia bia có thể mua được là:", tong_so_chai)
```
